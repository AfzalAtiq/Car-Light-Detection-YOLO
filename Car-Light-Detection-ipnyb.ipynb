```python
# =========================
# Install dependencies
# =========================
!pip install ultralytics opencv-python-headless

# =========================
# Imports
# =========================
import cv2
from ultralytics import YOLO
from IPython.display import HTML
from base64 import b64encode

# =========================
# Load YOLO model
# =========================
BEST_WEIGHTS = "/content/drive/MyDrive/brake detection dataset /yolo_runs/train_detect2/weights/best.pt"
model = YOLO(BEST_WEIGHTS)

print("✅ Model class names:", model.names)

# Keep only these classes
allowed_classes = ["light-on", "light-off"]

# =========================
# Input / Output paths
# =========================
video_path = "/content/drive/MyDrive/brake detection dataset/test video/Traffic IP Box Camera Made in Taiwan - Night with Color.mp4"
output_path = "/content/drive/MyDrive/brake detection dataset/test video/detected3.avi"

# =========================
# Open video
# =========================
cap = cv2.VideoCapture(video_path)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # mp4 codec
fps = int(cap.get(cv2.CAP_PROP_FPS))
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Define output video writer (no FPS reduction this time)
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    frame_count += 1

    # =========================
    # Run YOLO inference with very low conf
    # =========================
    results = model(frame, conf=0.3, verbose=False)
    detections = results[0]

    final_boxes = []
    if detections.boxes:
        cls_ids = detections.boxes.cls.int().cpu().tolist()
        keep = [i for i, c in enumerate(cls_ids) if model.names[c] in allowed_classes]
        if keep:
            detections.boxes = detections.boxes[keep]
            final_boxes = detections.boxes

    # If still nothing, take top-1 anyway
    if (not final_boxes) and detections.boxes:
        best_idx = detections.boxes.conf.argmax().item()
        detections.boxes = detections.boxes[[best_idx]]

    # =========================
    # Annotate frame (green = on, red = off)
    # =========================
    annotated_frame = frame.copy()
    if detections.boxes and len(detections.boxes) > 0:
        for box in detections.boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())
            cls_id = int(box.cls.item())
            conf = float(box.conf.item())
            label = model.names[cls_id]

            # Show all light detections with conf >= 0.3
            if conf >= 0.3 and label in ["light-on", "light-off"]:
                # Green for light-on, Red for light-off
                color = (0, 255, 0) if label == "light-on" else (0, 0, 255)

                # Draw box + label
                cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), color, 3)
                cv2.putText(
                    annotated_frame,
                    f"{label} {conf:.2f}",
                    (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.9,
                    color,
                    2,
                    cv2.LINE_AA
                )

    out.write(annotated_frame)

cap.release()
out.release()

# =========================
# Inline Preview in Colab
# =========================
mp4 = open(output_path,'rb').read()
data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
display(HTML(f"""
<video width=600 controls>
      <source src="{data_url}" type="video/mp4">
</video>
"""))
```
